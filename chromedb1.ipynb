```python
import csv

documents = []
metadatas = []
ids = []

# Mapping category numbers to their corresponding labels
category_map = {
    "1": "Business & Finance / Society & Culture",
    "2": "Computers & Internet / Technology",
    "3": "Entertainment & Music / Media",
    "4": "Family & Relationships",
    "5": "General Knowledge / Education / Trivia",
    "6": "Health & Medical",
    "7": "Science & Mathematics"
}

with open("newtrain.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    
    for row_id, row in enumerate(reader, start=1):
        cat_num = row["Category"].strip()
        text = row["Text"].strip()
        
        documents.append(text)
        metadatas.append({"category": category_map.get(cat_num, "Unknown")})
        ids.append(str(row_id))
```

```python
import chromadb
from chromadb.utils import embedding_functions
chroma_client = chromadb.Client()
```

```python
sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name="all-mpnet-base-v2")
collection = chroma_client.create_collection(name="my_collection", embedding_function=sentence_transformer_ef)
```

```python
collection.add(
    ids=ids,
    documents=documents,
    metadatas=metadatas
)
```

```python
chroma_client.delete_collection(name="my_collection")
```

```python
results = collection.query(
    query_texts=["who is more powerful spiderman or batman"], # Chroma will embed this for you
    n_results=2, 
    include=['distances','metadatas','documents']
)
results
```
