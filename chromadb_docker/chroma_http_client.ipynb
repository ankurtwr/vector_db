```python
import chromadb
from chromadb.utils import embedding_functions
from chromadb.config import Settings
```

```python
chroma_client = chromadb.HttpClient(host='localhost',port=8800,)
```

```python
sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name="all-mpnet-base-v2")
collection = chroma_client.get_or_create_collection(name="my_collection", embedding_function=sentence_transformer_ef)
```

```python
import csv

with open('menu_items_650.csv') as file:
    lines = csv.reader(file)
    
    documents= []
    metadatas= []
    ids= []
    id= 1 
    
    for i,line in enumerate(lines):
        if i==0:
            continue
        documents.append(line[1])
        metadatas.append({'id':line[0]})
        ids.append(str(id))
        id+=1
```

```python
collection.add(
    documents=documents,
    metadatas=metadatas,
    ids=ids
)
```

```python
print(collection.count())
```

```python
results = collection.query(
    query_texts=["roti"],
    n_results=5,
    include=['distances','documents','metadatas']
)
print(results['documents'])

results = collection.query(
    query_texts=["raita"],
    n_results=5,
    include=['distances','documents','metadatas']
)
print(results['documents'])

results = collection.query(
    query_texts=["idli"],
    n_results=5,
    include=['distances','documents','metadatas']
)
print(results['documents'])
```
